```python
import numpy as np
import matplotlib.pyplot as plt

# Definición de vectores de tiempo y frecuencia (omega)
t = np.linspace(-1, 5, 1000)
w = np.linspace(-15, 15, 1000)

# Señal en el tiempo
u_t_minus_1 = np.heaviside(t - 1, 1)
x1_t = np.exp(-3 * t) * u_t_minus_1

# Espectro de Magnitud |X1(w)|
X1_w = np.exp(-(3 + 1j * w)) / (3 + 1j * w)
X1_mag = np.abs(X1_w)

plt.figure(figsize=(12, 10))

# Gráfica Señal 1 en el tiempo
plt.subplot(2, 2, 1)
plt.plot(t, x1_t, 'b-', label='$x_1(t) = e^{-3t}u(t-1)$')
plt.title('Señal 1')
plt.xlabel('Tiempo (t)')
plt.ylabel('Amplitud')
plt.grid(True)
plt.legend()

# Gráfica Magnitud Transformada 1
plt.subplot(2, 2, 2)
plt.plot(w, X1_mag, 'r-', label='$|X_1(\omega)|$')
plt.title('Transformada de Fourier de la Señal 1')
plt.xlabel('Frecuencia ($\omega$) [rad/s]')
plt.ylabel('Magnitud')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parámetros del tiempo y señales
fs = 20000  # Frecuencia de muestreo (20 kHz)
t = np.arange(0, 0.03, 1/fs)  # 30 ms

# Componentes
x0 = 10 * np.sin(2 * np.pi * 100 * t + np.deg2rad(50))     # 100 Hz
r1 = 5 * np.sin(2 * np.pi * 1000 * t + np.deg2rad(10))     # 1000 Hz
r2 = 5 * np.sin(2 * np.pi * 1500 * t + np.deg2rad(20))     # 1500 Hz
r3 = 5 * np.sin(2 * np.pi * 2000 * t + np.deg2rad(30))     # 2000 Hz

# Entrada total
x_in = x0 + r1 + r2 + r3

# 2. Respuesta del Filtro RC en el dominio del tiempo 
#filtrado analítico
fc = 205
H_100  = 1 / np.sqrt(1 + (100/fc)**2)
H_1000 = 1 / np.sqrt(1 + (1000/fc)**2)
H_1500 = 1 / np.sqrt(1 + (1500/fc)**2)
H_2000 = 1 / np.sqrt(1 + (2000/fc)**2)

# Desfases teóricos: phi = -arctan(f / fc)
phi_100  = -np.arctan(100/fc)
phi_1000 = -np.arctan(1000/fc)
phi_1500 = -np.arctan(1500/fc)
phi_2000 = -np.arctan(2000/fc)

# Señal de salida filtrada
y_out = (10 * H_100  * np.sin(2 * np.pi * 100 * t  + np.deg2rad(50) + 
         phi_100) +
         5  * H_1000 * np.sin(2 * np.pi * 1000 * t + np.deg2rad(10) + 
         phi_1000) +
         5  * H_1500 * np.sin(2 * np.pi * 1500 * t + np.deg2rad(20) + 
         phi_1500) +
         5  * H_2000 * np.sin(2 * np.pi * 2000 * t + np.deg2rad(30) + 
         phi_2000))

# 3. Creación de figura con dos subplots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))

# Subplot 1: Dominio del tiempo
ax1.plot(t * 1000, x_in, label='Entrada: Señal ruidosa $x(t)$', 
color='gray', alpha=0.7)
ax1.plot(t * 1000, y_out, label='Salida: Señal recuperada $y(t)$', 
color='red', linewidth=2)
ax1.set_title('Comparación en el Dominio del Tiempo (0 a 30 ms)')
ax1.set_xlabel('Tiempo [ms]')
ax1.set_ylabel('Amplitud [V]')
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend(loc='upper right')

# Subplot 2: Diagrama de Bode (Respuesta en Frecuencia)
freqs = np.logspace(1, 4, 500)
ganancia = 1 / np.sqrt(1 + (freqs / fc)**2)

ax2.semilogx(freqs, ganancia, color='blue', linewidth=2, 
label='Respuesta $|H(f)|$')
ax2.plot(100, H_100, 'go', label=f'100 Hz (Ganancia: {H_100:.3f})')
ax2.plot(fc, 1/np.sqrt(2), 'ro', label=f'fc = {fc} Hz (-3dB)')
ax2.plot(1000, H_1000, 'mo', label=f'1000 Hz (Ganancia: {H_1000:.3f})')

ax2.set_title('Respuesta en Frecuencia del Filtro (Diagrama de Bode)')
ax2.set_xlabel('Frecuencia [Hz]')
ax2.set_ylabel('Ganancia Magnitude')
ax2.grid(True, which='both', linestyle='--', alpha=0.5)
ax2.legend(loc='lower left')

plt.tight_layout()
plt.show()
```

```python
# Filtros pasa-altos de primer orden en cascada (PROB. FILTROS)
import numpy as np
import matplotlib.pyplot as plt

# 1. Parametros del tiempo y senales
fs = 20000   # Frecuencia de muestreo (20 kHz)
t = np.arange(0, 0.03, 1/fs)   # 30 ms

# Componentes originales de la senal contaminada
x0 = 10 * np.sin(2 * np.pi * 100 * t + np.deg2rad(50))      # 100 Hz
r1 = 5 * np.sin(2 * np.pi * 1000 * t + np.deg2rad(10))      # 1000 Hz
r2 = 5 * np.sin(2 * np.pi * 1500 * t + np.deg2rad(20))      # 1500 Hz
r3 = 5 * np.sin(2 * np.pi * 2000 * t + np.deg2rad(30))      # 2000 Hz

# Entrada total
x_in = x0 + r1 + r2 + r3

# 2. Respuesta de Filtro Pasa-Altos (2 Etapas RC) con fc limitante
fc = 4500    # Frecuencia de corte calculada para atenuar 1500Hz al 90%

def ganancia_etapa_hp(f, fc):
    return 1 / np.sqrt(1 + (fc / f)**2)

def desfase_etapa_hp(f, fc):
    return np.arctan(fc / f)

def ganancia_total_hp2(f, fc):
    return ganancia_etapa_hp(f, fc)**2

def desfase_total_hp2(f, fc):
    return 2 * desfase_etapa_hp(f, fc)

freqs_eval = [100, 1000, 1500, 2000]
H_vals = [ganancia_total_hp2(f, fc) for f in freqs_eval]
phi_vals = [desfase_total_hp2(f, fc) for f in freqs_eval]

H_100, H_1000, H_1500, H_2000 = H_vals
phi_100, phi_1000, phi_1500, phi_2000 = phi_vals

# Senal de salida filtrada (Notar que H_2000 sera muy bajo)
y_out = (10 * H_100   * np.sin(2 * np.pi * 100 * t   + np.deg2rad(50) + phi_100) +
         5  * H_1000  * np.sin(2 * np.pi * 1000 * t  + np.deg2rad(10) + phi_1000) +
         5  * H_1500  * np.sin(2 * np.pi * 1500 * t  + np.deg2rad(20) + phi_1500) +
         5  * H_2000  * np.sin(2 * np.pi * 2000 * t  + np.deg2rad(30) + phi_2000))

# 3. Graficas
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8))

ax1.plot(t * 1000, x_in, label='Entrada: Senal contaminada $x(t)$', color='gray', alpha=0.7)
ax1.plot(t * 1000, y_out, label='Salida: Senal atenuada $y(t)$', color='red', linewidth=2)
ax1.set_title('Filtro Pasa-Altos (2 Etapas RC, fc=4500Hz)')
ax1.set_xlabel('Tiempo [ms]')
ax1.set_ylabel('Amplitud [V]')
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend(loc='upper right')

freqs = np.logspace(1, 4.2, 500)
ganancia_bode = [ganancia_total_hp2(f, fc) for f in freqs]

ax2.semilogx(freqs, ganancia_bode, color='blue', linewidth=2, label='Respuesta 2 Etapas Cascada')
ax2.plot(100, H_100, 'ko', label=f'100 Hz')
ax2.plot(1000, H_1000, 'mo', label=f'1000 Hz')
ax2.plot(1500, H_1500, 'go', label=f'1500 Hz (Ganancia: {H_1500:.3f})')
ax2.plot(fc, (1/np.sqrt(2))**2, 'ro', label=f'fc = {fc} Hz (-6dB)')
ax2.plot(2000, H_2000, 'bo', label=f'2000 Hz (Ganancia: {H_2000:.3f})')

ax2.set_title('Diagrama de Bode evidenciando la atenuacion de 2 kHz')
ax2.set_xlabel('Frecuencia [Hz]')
ax2.set_ylabel('Ganancia Magnitude')
ax2.grid(True, which='both', linestyle='--', alpha=0.5)
ax2.legend(loc='upper left')

plt.tight_layout()
plt.show()
```
